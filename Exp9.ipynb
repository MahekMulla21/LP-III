```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):            # function
    return (x + 3) ** 2

def df(x):           # derivative dy/dx = 2(x+3)
    return 2 * (x + 3)
```

```python
x = 2                 # starting point
learning_rate = 0.1
epochs = 50
tolerance = 1e-6
history = [x]

for i in range(epochs):
    grad = df(x)
    x_new = x - learning_rate * grad
    print(f"Iteration {i+1}: x = {x_new:.6f}, y = {f(x_new):.6f}")
    history.append(x_new)
    if abs(x_new - x) < tolerance:
        break
    x = x_new

print("\nLocal minimum occurs at x =", round(x, 4))
print("Minimum value of y =", round(f(x), 6))
```

```python
xs = np.linspace(-8, 4, 200)
plt.plot(xs, f(xs), label="y = (x+3)^2")
plt.scatter(history, [f(h) for h in history], color="red", s=20, label="Gradient descent steps")
plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.grid(True)
plt.show()
```
